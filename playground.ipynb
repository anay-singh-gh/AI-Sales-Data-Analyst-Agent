## Ask questions

Call `ask("your question")` — it returns a pandas DataFrame. Conversation history is kept, so follow-up questions work naturally. Call `reset()` any time to start a fresh conversation.

The generated SQL is hidden by default — pass `ask(question, show_sql=True)` if you want to see it.

If your question explicitly asks for a chart ("plot", "graph", "visualize", "show me a bar chart of...") the agent will also render one automatically. Plain questions just return the table — no chart is forced on you.

```python
ask("Who are the top 5 customers by total revenue?")
```

```python
ask("Which products have sold the most units?")
```

```python
# Follow-up question — uses the previous conversation as context
ask("Which of those has the highest revenue?")
```

```python
reset()  # clear conversation history and start over
```

```python
# Explicit chart request — the agent decides to visualize because you asked for one
ask("Show me a bar chart of the top 5 customers by total revenue")
```
